```python
import pandas as pd
import matplotlib.pyplot as plt

# Load data
df = pd.read_csv("shrinath sale.csv", sep="\t")

print("===== DATA PREVIEW =====")
print(df.head())

# =========================
# BASIC STATISTICS
# =========================
print("\n===== STATISTICS =====")
print("Mean Sales:", df["Sales"].mean())
print("Median Sales:", df["Sales"].median())
print("Mode Sales:", df["Sales"].mode()[0])

# =========================
# ADD NEW COLUMN
# =========================
df["Profit %"] = (df["Profit"] / df["Sales"]) * 100

# =========================
# DELETE ROW & COLUMN
# =========================
df = df.drop(0)  # delete first row
df = df.drop("Customer Name", axis=1)  # delete column

# =========================
# FILTER DATA
# =========================
filtered = df[df["Sales"] > 500]
print("\nFiltered Data (Sales > 500):")
print(filtered.head())

# =========================
# SORT DATA
# =========================
sorted_df = df.sort_values(by="Sales", ascending=False)
print("\nTop Sales Data:")
print(sorted_df.head())

# =========================
# GROUPING
# =========================
category_sales = df.groupby("Category")["Sales"].sum()
print("\nCategory Wise Sales:")
print(category_sales)

# =========================
# GRAPHS
# =========================

# Bar Chart
category_sales.plot(kind="bar")
plt.title("Category Wise Sales")
plt.xlabel("Category")
plt.ylabel("Sales")
plt.show()

# Line Chart (Date vs Sales)
df["Order Date"] = pd.to_datetime(df["Order Date"], dayfirst=True)
daily_sales = df.groupby("Order Date")["Sales"].sum()

daily_sales.plot()
plt.title("Daily Sales Trend")
plt.xlabel("Date")
plt.ylabel("Sales")
plt.show()
```

```python
import pandas as pd

# Load data
df = pd.read_csv("shrinath sale.csv", sep="\t")

print("===== DATA PREVIEW =====")
print(df.head())

# =========================
# BASIC INFO
# =========================
print("\nShape:", df.shape)
print("\nColumns:", df.columns)
print("\nData Types:\n", df.dtypes)

# =========================
# MISSING VALUES
# =========================
print("\nMissing Values:\n", df.isnull().sum())

# Fill missing values
df["Sales"] = df["Sales"].fillna(0)

# =========================
# RENAME COLUMN
# =========================
df = df.rename(columns={"Sales": "Total_Sales"})

# =========================
# UNIQUE & COUNT
# =========================
print("\nUnique Categories:", df["Category"].unique())
print("\nCategory Count:\n", df["Category"].value_counts())

# =========================
# SELECT DATA
# =========================
print("\nSelected Columns:\n", df[["Total_Sales", "Profit"]])

print("\nFirst 5 Rows using iloc:\n", df.iloc[0:5])

print("\nSales > 1000:\n", df.loc[df["Total_Sales"] > 1000])

# =========================
# MULTIPLE FILTER
# =========================
print("\nSales > 500 and Profit > 100:\n",
      df[(df["Total_Sales"] > 500) & (df["Profit"] > 100)])

# =========================
# GROUP BY
# =========================
print("\nGroup By Category & Region:\n",
      df.groupby(["Category", "Region"])["Total_Sales"].sum())

# =========================
# PIVOT TABLE
# =========================
pivot = pd.pivot_table(df, values="Total_Sales",
                       index="Category", columns="Region", aggfunc="sum")
print("\nPivot Table:\n", pivot)

# =========================
# MAX / MIN
# =========================
print("\nMax Sales:", df["Total_Sales"].max())
print("Min Sales:", df["Total_Sales"].min())

# =========================
# COUNT ROWS
# =========================
print("\nTotal Rows:", len(df))

# =========================
# DROP DUPLICATES
# =========================
df = df.drop_duplicates()

# =========================
# SORT DATA
# =========================
df = df.sort_values(by=["Total_Sales", "Profit"], ascending=False)

# =========================
# APPLY FUNCTION
# =========================
df["Double Sales"] = df["Total_Sales"].apply(lambda x: x * 2)

# =========================
# DELETE ROW & COLUMN
# =========================
df = df.drop(0)  # delete first row
df = df.drop("Customer Name", axis=1)

# =========================
# SAVE FILE
# =========================
df.to_csv("output.csv", index=False)

print("\n✅ Process Completed & File Saved!")
```
